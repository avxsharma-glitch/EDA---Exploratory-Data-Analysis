## Dataset Understanding

The dataset contains information about Airbnb listings, including location,
price, room type, host details, customer reviews, and property availability.

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
df.columns
```

```python
missing_values = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_values.sort_values(
    by="Missing Values",
    ascending=False
)
```

```python
df.columns.tolist()
```

```python
df.drop(columns=["license"], inplace=True)
```

```python
df["price"].head()
```

```python
df["price"] = (
    df["price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["price"] = pd.to_numeric(
    df["price"],
    errors="coerce"
)
```

```python
df["price"].describe()
```

```python
df["service_fee"] = (
    df["service_fee"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
)

df["service_fee"] = pd.to_numeric(
    df["service_fee"],
    errors="coerce"
)
```

```python
df["price"] = df["price"].fillna(df["price"].median())

df["service_fee"] = df["service_fee"].fillna(
    df["service_fee"].median()
)
```

```python
df["house_rules"] = df["house_rules"].fillna(
    "Not Specified"
)
```

```python
df["reviews_per_month"] = df["reviews_per_month"].fillna(0)
```

```python
df.to_csv(
    "cleaned_airbnb_data.csv",
    index=False
)
```

```python
df["room_type"].value_counts()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="room_type"
)

plt.title("Distribution of Airbnb Room Types")
plt.xlabel("Room Type")
plt.ylabel("Number of Listings")

plt.xticks(rotation=20)

plt.show()
```

```python
top_locations = (
    df["neighbourhood"]
    .value_counts()
    .head(10)
)

top_locations
```

```python
plt.figure(figsize=(10, 6))

top_locations.sort_values().plot(
    kind="barh"
)

plt.title("Top 10 Neighbourhoods by Number of Listings")
plt.xlabel("Number of Listings")
plt.ylabel("Neighbourhood")

plt.show()
```

```python
avg_price_room = (
    df.groupby("room_type")["price"]
    .mean()
    .sort_values(ascending=False)
)

avg_price_room
```

```python
plt.figure(figsize=(8, 5))

avg_price_room.plot(
    kind="bar"
)

plt.title("Average Price by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Price")

plt.xticks(rotation=20)

plt.show()
```

```python
avg_price_location = (
    df.groupby("neighbourhood_group")["price"]
    .mean()
    .sort_values(ascending=False)
)

avg_price_location
```

```python
plt.figure(figsize=(10, 6))

avg_price_location.plot(
    kind="bar"
)

plt.title("Average Airbnb Price by Neighbourhood Group")
plt.xlabel("Neighbourhood Group")
plt.ylabel("Average Price")

plt.xticks(rotation=20)

plt.show()
```

```python
review_data = df[
    ["price", "number_of_reviews"]
].dropna()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=review_data,
    x="price",
    y="number_of_reviews"
)

plt.title("Relationship Between Price and Number of Reviews")

plt.show()
```

```python
filtered_data = review_data[
    review_data["price"] < review_data["price"].quantile(0.95)
]
```

```python
availability = (
    df.groupby("room_type")["availability_365"]
    .mean()
    .sort_values(ascending=False)
)

availability
```

```python
plt.figure(figsize=(8, 5))

availability.plot(
    kind="bar"
)

plt.title("Average Availability by Room Type")
plt.xlabel("Room Type")
plt.ylabel("Average Available Days")

plt.xticks(rotation=20)

plt.show()
```

```python
host_analysis = df.groupby(
    "host_identity_verified"
).agg({
    "price": "mean",
    "number_of_reviews": "mean",
    "review_rate_number": "mean"
})

host_analysis
```

```python
# Key Insights from the Analysis

print("1. Total Airbnb Listings:", len(df))

print("\n2. Most Common Room Type:")
print(df["room_type"].value_counts().idxmax())

print("\n3. Neighbourhood with Highest Number of Listings:")
print(df["neighbourhood"].value_counts().idxmax())

print("\n4. Neighbourhood Group with Highest Average Price:")
print(
    df.groupby("neighbourhood_group")["price"]
    .mean()
    .idxmax()
)

print("\n5. Room Type with Highest Average Price:")
print(
    df.groupby("room_type")["price"]
    .mean()
    .idxmax()
)

print("\n6. Room Type with Highest Average Availability:")
print(
    df.groupby("room_type")["availability_365"]
    .mean()
    .idxmax()
)

print("\n7. Host Verification Analysis:")
print(
    df.groupby("host_identity_verified")
    .agg(
        Average_Price=("price", "mean"),
        Average_Reviews=("number_of_reviews", "mean")
    )
)
```
