```python
df = pd.read_csv("1776250375-P2-Healthcare Analytics for Doctor Visits.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

df.info()
```

```python
# Remove unnecessary index column
df = df.drop(columns=["Unnamed: 0"])

# Check the updated dataset
df.head()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df.describe()
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts())
```

```python
# Select numeric columns
numeric_df = df.select_dtypes(include=["int64", "float64"])

# Calculate correlation
correlation = numeric_df.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of Healthcare Variables")

plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Create images folder
os.makedirs("images", exist_ok=True)


# 1. Distribution of Doctor Visits
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x="visits", bins=30, kde=True)
plt.title("Distribution of Doctor Visits")
plt.xlabel("Number of Doctor Visits")
plt.ylabel("Number of Patients")
plt.savefig("images/doctor_visits_distribution.png", dpi=300, bbox_inches="tight")
plt.show()


# 2. Average Doctor Visits by Gender
plt.figure(figsize=(7, 5))
sns.barplot(data=df, x="gender", y="visits")
plt.title("Average Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Doctor Visits")
plt.savefig("images/visits_by_gender.png", dpi=300, bbox_inches="tight")
plt.show()


# 3. Age vs Doctor Visits
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="age", y="visits", alpha=0.5)
plt.title("Age vs Doctor Visits")
plt.xlabel("Age")
plt.ylabel("Number of Doctor Visits")
plt.savefig("images/age_vs_visits.png", dpi=300, bbox_inches="tight")
plt.show()


# 4. Illness vs Doctor Visits
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x="illness", y="visits")
plt.title("Average Doctor Visits by Number of Illnesses")
plt.xlabel("Number of Illnesses")
plt.ylabel("Average Doctor Visits")
plt.savefig("images/illness_vs_visits.png", dpi=300, bbox_inches="tight")
plt.show()


# 5. Health Status vs Doctor Visits
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x="health", y="visits")
plt.title("Average Doctor Visits by Health Status")
plt.xlabel("Health Status")
plt.ylabel("Average Doctor Visits")
plt.savefig("images/health_vs_visits.png", dpi=300, bbox_inches="tight")
plt.show()


# 6. Income vs Doctor Visits
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="income", y="visits", alpha=0.5)
plt.title("Income vs Doctor Visits")
plt.xlabel("Income")
plt.ylabel("Number of Doctor Visits")
plt.savefig("images/income_vs_visits.png", dpi=300, bbox_inches="tight")
plt.show()


# 7. Private Insurance vs Doctor Visits
plt.figure(figsize=(7, 5))
sns.barplot(data=df, x="private", y="visits")
plt.title("Average Doctor Visits by Private Insurance")
plt.xlabel("Private Insurance")
plt.ylabel("Average Doctor Visits")
plt.savefig("images/private_insurance_vs_visits.png", dpi=300, bbox_inches="tight")
plt.show()


# 8. Correlation Heatmap
numeric_df = df.select_dtypes(include=["int64", "float64"])
correlation = numeric_df.corr()

plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Healthcare Variables")
plt.savefig("images/correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()


print("All charts saved successfully!")

print("\nSaved files:")
print(os.listdir("images"))
```
